# Prompt Engineering

## Qué es prompt engineering
- Diseño de instrucciones claras para guiar al modelo.
- Mejora calidad, estabilidad y reproducibilidad.

## Estructura de un buen prompt
- Contexto y rol.
- Objetivo específico.
- Formato de salida.
- Restricciones o ejemplos si son necesarios.

```python
# Ejemplo de prompt mal estructurado
from ollama import Client
client = Client(host="http://localhost:11434")
bad = client.chat(model="llama3", messages=[{"role":"user","content":"Cuéntame cosas."}])
print(bad["message"]["content"])
```

```python
# Ejemplo de prompt bien estructurado con rol de sistema
messages = [
    {"role": "system", "content": "Eres un analista técnico conciso."},
    {"role": "user", "content": "Explica qué es un LLM en 4 puntos y da un ejemplo de uso."}
]
good = client.chat(model="llama3", messages=messages, options={"temperature": 0.2})
print(good["message"]["content"])
```
