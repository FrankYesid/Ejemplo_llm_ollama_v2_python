# Ejemplos de uso práctico

Incluye resumen, clasificación, generación de explicaciones técnicas, extracción de información y reescritura de contenido.

Cada ejemplo explica el problema, muestra el prompt y analiza la respuesta.

```python
# Configuración inicial y lectura de datos
from ollama import Client
from pathlib import Path
from scripts.prompts import prompt_resumen, prompt_clasificacion, prompt_extraccion, prompt_reescritura
from scripts.utils import safe_read_file

client = Client(host="http://localhost:11434")
texto = safe_read_file(Path("../data/textos_ejemplo.txt"))
print(texto[:200], "...")
```

## Resumen automático de texto
- Problema: obtener un resumen breve y fiel.
- Prompt: rol de sistema experto en resumen, salida en 5-7 líneas.
- Análisis: evaluar claridad, cobertura y ausencia de alucinaciones.

```python
res = client.chat(model="llama3", messages=prompt_resumen(texto))
print(res["message"]["content"])
```

## Clasificación de texto
- Problema: clasificar sentimiento en categorías.
- Prompt: salida restringida a una etiqueta.
- Análisis: verificar consistencia y formato exacto.

```python
etiquetas = ["positivo", "negativo", "neutral"]
res2 = client.chat(model="llama3", messages=prompt_clasificacion(texto, etiquetas))
print(res2["message"]["content"])
```

## Extracción de información
- Problema: extraer campos específicos.
- Prompt: salida JSON válida.
- Análisis: validar estructura y valores.

```python
campos = ["tema", "beneficios", "riesgos"]
res3 = client.chat(model="llama3", messages=prompt_extraccion(texto, campos))
print(res3["message"]["content"])
```

## Reescritura de contenido
- Problema: mejorar claridad y ajustar tono.
- Prompt: definir tono deseado.
- Análisis: conservar significado y mejorar estilo.

```python
res4 = client.chat(model="llama3", messages=prompt_reescritura(texto, tono="técnico y claro"))
print(res4["message"]["content"])
```
