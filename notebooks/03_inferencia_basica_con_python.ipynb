# Inferencia básica con Python

Este cuaderno muestra cómo usar el paquete `ollama` para realizar una inferencia básica con un modelo local.

## Pasos
- Conectar con el servidor de Ollama.
- Enviar un prompt simple.
- Ajustar parámetros básicos.

```python
# Conexión y primer prompt
from ollama import Client
client = Client(host="http://localhost:11434")
res = client.chat(model="llama3", messages=[{"role":"user","content":"Explica brevemente qué es un LLM."}])
print(res["message"]["content"])
```

```python
# Parámetros básicos del modelo
options = {"temperature": 0.2, "top_p": 0.9}
res2 = client.chat(model="llama3", messages=[{"role":"user","content":"Da 3 ventajas de ejecutar LLMs de forma local."}], options=options)
print(res2["message"]["content"])
```
