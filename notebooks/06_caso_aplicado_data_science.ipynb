# Caso aplicado a Data Science: Asistente EDA

## Contexto
Usaremos un LLM local como asistente para análisis exploratorio de datos (EDA), generación de hipótesis y documentación técnica del proyecto.

## Flujo
- Definir contexto del proyecto.
- Solicitar hipótesis y preguntas exploratorias.
- Pedir sugerencias de visualizaciones y pasos siguientes.
- Registrar reflexiones y documentación resumida.


```python
# Configuración y prompt EDA
from ollama import Client
from scripts.prompts import prompt_eda

client = Client(host="http://localhost:11434")
contexto = (
    "Proyecto de ventas minoristas con datos de transacciones, clientes y productos. "
    "Objetivo: explicar variaciones mensuales y recomendar acciones para mejorar el margen."
)
instrucciones = (
    "Incluye hipótesis cuantitativas y cualitativas, sugiere visualizaciones, "
    "posibles variables confusoras y un breve checklist de EDA reproducible."
)
messages = prompt_eda(contexto, instrucciones)
res = client.chat(model="llama3", messages=messages, options={"temperature": 0.2})
print(res["message"]["content"])
```

## Reflexiones finales
- El LLM ayuda a estructurar el análisis y documentación.
- No sustituye la validación estadística ni la ejecución de código analítico.
- Úsalo para ideación, planificación y comunicación técnica.
