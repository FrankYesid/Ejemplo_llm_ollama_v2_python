# Instalación y descarga de modelos con Ollama

## Verificación de instalación
- Asegúrate de tener Ollama instalado.
- Verifica con `ollama --version`.

## Descarga de modelos
- Usa `ollama pull <modelo>` para descargar modelos.
- Ejemplos: `llama3`, `mistral`.

## Recomendaciones de hardware
- Verifica memoria disponible y compatibilidad de GPU si aplica.
- Selecciona modelos acordes a tus recursos.

```python
# Comandos básicos usando subprocess (salida dependerá de tu sistema)
import subprocess
def run(cmd):
    try:
        out = subprocess.run(cmd, shell=True, capture_output=True, text=True)
        print(out.stdout or out.stderr)
    except Exception as e:
        print("Error:", e)

run("ollama --version")
run("ollama list")
```

```python
# Descarga de modelos (ejemplo)
run("ollama pull llama3")
run("ollama pull mistral")
```
