# Introducción a LLMs y Ollama

Este cuaderno presenta conceptos clave sobre los Modelos de Lenguaje Grandes (LLMs) y el rol de Ollama para ejecutarlos de forma local.

## Qué es un LLM
- Un LLM es un modelo de IA entrenado para comprender y generar texto.
- Habilita tareas como resumen, clasificación, extracción y generación.

## Ejecución local
- Ejecutar un modelo localmente evita enviar datos a servicios externos.
- Beneficios: privacidad, costos y control del entorno.

## Ollama en el ecosistema
- Ollama permite descargar, gestionar y servir LLMs localmente.
- Se integra con Python mediante el paquete `ollama`.
